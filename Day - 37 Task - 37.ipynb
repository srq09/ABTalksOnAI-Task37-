```python
import os
import numpy as np
from dotenv import load_dotenv
from openai import OpenAI

# ==============================
# LOAD ENV VARIABLES
# ==============================

load_dotenv()

api_key = os.getenv("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")


# ==============================
# OPENAI CLIENT
# ==============================

client = OpenAI(api_key= "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")

# ==============================
# SAMPLE DOCUMENTS
# ==============================

documents = [
    "Artificial Intelligence enables machines to think like humans.",
    "Machine Learning is a subset of Artificial Intelligence.",
    "Deep Learning uses neural networks with multiple layers.",
    "Python is a popular programming language for AI.",
    "OpenAI provides advanced AI APIs and models."
]

# ==============================
# GENERATE EMBEDDING FUNCTION
# ==============================

def generate_embedding(text):

    response = client.embeddings.create(
        model="text-embedding-3-small",
        input=text
    )

    embedding = response.data[0].embedding

    return embedding

# ==============================
# COSINE SIMILARITY FUNCTION
# ==============================

def cosine_similarity(vec1, vec2):

    vec1 = np.array(vec1)
    vec2 = np.array(vec2)

    similarity = np.dot(vec1, vec2) / (
        np.linalg.norm(vec1) * np.linalg.norm(vec2)
    )

    return similarity

# ==============================
# GENERATE DOCUMENT EMBEDDINGS
# ==============================

print("\nGenerating embeddings...\n")

embedded_documents = []

for doc in documents:

    embedding = generate_embedding(doc)

    embedded_documents.append({
        "text": doc,
        "embedding": embedding
    })

    print(f"Embedding generated for: {doc}")

# ==============================
# USER QUERY
# ==============================

query = input("\nEnter your search query: ")

query_embedding = generate_embedding(query)

# ==============================
# CALCULATE SIMILARITIES
# ==============================

results = []

for item in embedded_documents:

    similarity = cosine_similarity(
        query_embedding,
        item["embedding"]
    )

    results.append({
        "text": item["text"],
        "score": similarity
    })

# ==============================
# SORT RESULTS
# ==============================

results = sorted(
    results,
    key=lambda x: x["score"],
    reverse=True
)

# ==============================
# DISPLAY RESULTS
# ==============================

print("\n========== SEARCH RESULTS ==========\n")

for rank, result in enumerate(results, start=1):

    print(f"Rank #{rank}")
    print(f"Similarity Score: {result['score']:.4f}")
    print(f"Document: {result['text']}")
    print("-" * 50)
```
